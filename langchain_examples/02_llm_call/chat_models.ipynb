```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
load_dotenv()
model = ChatOpenAI(
    model="qwen-plus",
    api_key=os.getenv("ALIYUN_API_KEY"),
    base_url="https://dashscope.aliyuncs.com/compatible-mode/v1",
)
model
```

```python
from http.client import responses
import sys
import io
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
#  手动实现一个多轮对话的内容
# 实例化一个模型

messages  = [
    SystemMessage("你是一个有用的助手，可以帮助我完成一些任务。"),
    HumanMessage("你好，我是Luke，很高兴认识你。")
]
responses = model.invoke(messages)
print(responses.content)
```

```python
new_message = [HumanMessage("我的名字是什么？")]
res = model.invoke(new_message)
print(res.content)
```

```python
messages.append(AIMessage(content=responses.content))
messages.append(HumanMessage("我的名字是什么？"))
print(messages)
res = model.invoke(messages)
print(res.content)
```
