```python
# python中的反射机制
class Student():
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age


    def study(self):
        print(f"{self.name} is studying")

    def sleep(self):
        print(f"{self.name} is sleeping")
```

```python
jungeng = Student("俊耿", 22)

if hasattr(jungeng, "height"):
    print(jungeng.height)
else:
    print("没有height属性")


if hasattr(jungeng, "play_game"):
    jungeng.play_game()

# 动态的增加属性
if not hasattr(jungeng, "height"):
    setattr(jungeng, "height", "1.85")

print(jungeng.height)

print(getattr(jungeng, "height"))
```
